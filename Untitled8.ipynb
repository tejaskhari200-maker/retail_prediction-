```python

import pandas as pd
import numpy as np
from scipy.stats import t

results = pd.read_csv("/content/reports/baseline_results.csv")

# Keep the five comparison algorithms
results = results[
    results["model"] != "Reference_MedianPredictor"
].copy()

metrics = [
    "MAE", "RMSE", "R2",
    "train_time_s", "inference_time_s"
]

summary = results.groupby(
    ["scenario", "model"]
)[metrics].agg(["mean", "std", "count"])

# Calculate 95% confidence intervals for each mean
rows = []

for (scenario, model), group in results.groupby(
    ["scenario", "model"]
):
    row = {"scenario": scenario, "model": model}

    for metric in metrics:
        values = group[metric].dropna()
        n = len(values)
        mean = values.mean()
        std = values.std(ddof=1)

        if n > 1:
            margin = t.ppf(0.975, n - 1) * std / np.sqrt(n)
        else:
            margin = np.nan

        row[f"{metric}_mean"] = mean
        row[f"{metric}_std"] = std
        row[f"{metric}_CI_low"] = mean - margin
        row[f"{metric}_CI_high"] = mean + margin

    row["runs"] = n
    rows.append(row)

comparison = pd.DataFrame(rows).round(3)

display(comparison)

comparison.to_csv(
    "/content/reports/model_comparison_with_CI.csv",
    index=False
)

print("Saved model_comparison_with_CI.csv")
```

```python

import json
import time
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor
from common import make_pipeline, TARGET

df = pd.read_csv("/content/sales_data_validated.csv").set_index("Row ID")

with open("/content/splits.json") as f:
    splits = json.load(f)

def create_models(seed):
    return {
        "LinearRegression": LinearRegression(),
        "RandomForest": RandomForestRegressor(
            n_estimators=100, random_state=seed, n_jobs=-1
        ),
        "XGBoost": XGBRegressor(
            n_estimators=100, max_depth=6, learning_rate=0.1,
            random_state=seed, n_jobs=-1,
            objective="reg:squarederror"
        )
    }

all_results = []

for seed in [42, 43, 44, 45, 46]:
    print(f"\nRunning seed {seed}...")

    split = splits[f"order_grouped_seed{seed}"]
    train_df = df.loc[split["train_row_ids"]].copy()
    test_df = df.loc[split["test_row_ids"]].copy()

    X_train = train_df.drop(columns=[TARGET])
    y_train = train_df[TARGET]
    X_test = test_df.drop(columns=[TARGET])
    y_test = test_df[TARGET]
    groups = train_df["Order ID"].to_numpy()

    names = list(create_models(seed).keys())
    oof = np.zeros((len(X_train), len(names)))

    cv = GroupKFold(n_splits=5)

    for fit_idx, val_idx in cv.split(X_train, y_train, groups):
        for j, (name, model) in enumerate(create_models(seed).items()):
            pipe = make_pipeline(model, use_shipping_cost=False)
            pipe.fit(X_train.iloc[fit_idx], y_train.iloc[fit_idx])
            oof[val_idx, j] = pipe.predict(X_train.iloc[val_idx])

    meta = Ridge(alpha=1.0)
    meta.fit(oof, y_train)

    test_preds = np.zeros((len(X_test), len(names)))
    for j, (name, model) in enumerate(create_models(seed).items()):
        pipe = make_pipeline(model, use_shipping_cost=False)
        pipe.fit(X_train, y_train)
        test_preds[:, j] = pipe.predict(X_test)

    predictions = {
        "RandomForest": test_preds[:, names.index("RandomForest")],
        "StackingRegressor": meta.predict(test_preds)
    }

    for model_name, pred in predictions.items():
        all_results.append({
            "seed": seed,
            "model": model_name,
            "MAE": mean_absolute_error(y_test, pred),
            "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
            "R2": r2_score(y_test, pred)
        })

results_df = pd.DataFrame(all_results).round(4)
display(results_df)

summary = results_df.groupby("model")[["MAE", "RMSE", "R2"]].agg(
    ["mean", "std"]
).round(4)

print("\nFive-seed summary (mean and standard deviation):")
display(summary)

results_df.to_csv(
    "/content/reports/stacking_five_seeds_no_shipping.csv",
    index=False
)
print("Saved stacking_five_seeds_no_shipping.csv")
```

```python

import pandas as pd
import numpy as np
from scipy.stats import t

baseline = pd.read_csv("/content/reports/baseline_results.csv")
stacking = pd.read_csv(
    "/content/reports/stacking_five_seeds_no_shipping.csv"
)

# Random Forest baseline from the same no-shipping scenario
rf = baseline[
    (baseline["model"] == "RandomForest") &
    (baseline["scenario"] == "without_shipping_cost")
][["seed", "MAE", "RMSE", "R2"]].copy()

stack = stacking[
    stacking["model"] == "StackingRegressor"
][["seed", "MAE", "RMSE", "R2"]].copy()

paired = rf.merge(stack, on="seed", suffixes=("_RF", "_Stack"))

# Positive values mean stacking improved the metric
paired["MAE_improvement"] = paired["MAE_RF"] - paired["MAE_Stack"]
paired["RMSE_improvement"] = paired["RMSE_RF"] - paired["RMSE_Stack"]
paired["R2_improvement"] = paired["R2_Stack"] - paired["R2_RF"]

print("Per-seed paired improvements (positive = stacking better):")
display(paired[[
    "seed", "MAE_improvement",
    "RMSE_improvement", "R2_improvement"
]].round(4))

# Descriptive 95% t-intervals for mean paired improvements
summary_rows = []

for metric in [
    "MAE_improvement", "RMSE_improvement", "R2_improvement"
]:
    values = paired[metric].dropna()
    n = len(values)
    mean = values.mean()
    sd = values.std(ddof=1)
    margin = t.ppf(0.975, n - 1) * sd / np.sqrt(n)

    summary_rows.append({
        "Metric": metric,
        "Mean improvement": mean,
        "95% CI lower": mean - margin,
        "95% CI upper": mean + margin,
        "Seeds improved": int((values > 0).sum()),
        "Seeds evaluated": n
    })

paired_summary = pd.DataFrame(summary_rows).round(4)
print("Summary across paired seeds:")
display(paired_summary)

paired_summary.to_csv(
    "/content/reports/stacking_paired_improvement.csv",
    index=False
)

print("Saved stacking_paired_improvement.csv")
```

```python

import json
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from common import make_pipeline, TARGET

df = pd.read_csv("/content/sales_data_validated.csv").set_index("Row ID")

with open("/content/splits.json") as f:
    splits = json.load(f)

split = splits["order_grouped_seed42"]
train_df = df.loc[split["train_row_ids"]].copy()
test_df = df.loc[split["test_row_ids"]].copy()

X_test = test_df.drop(columns=[TARGET])
y_test = test_df[TARGET]

# Sample whole orders to avoid splitting an order across subsets
rng = np.random.default_rng(42)
orders = train_df["Order ID"].drop_duplicates().to_numpy()
rng.shuffle(orders)

results = []

for fraction in [0.50, 0.75, 1.00]:
    selected_count = max(1, int(len(orders) * fraction))
    selected_orders = set(orders[:selected_count])

    subset = train_df[train_df["Order ID"].isin(selected_orders)]
    X_train = subset.drop(columns=[TARGET])
    y_train = subset[TARGET]

    model = make_pipeline(
        RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),
        use_shipping_cost=False
    )

    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    results.append({
        "Training fraction": fraction,
        "Training rows": len(subset),
        "Training orders": selected_count,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions)
    })

robustness = pd.DataFrame(results).round(4)
display(robustness)

robustness.to_csv(
    "/content/reports/training_size_robustness.csv",
    index=False
)

print("Saved training_size_robustness.csv")
```

```python

import json
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor
from common import make_pipeline, TARGET

df = pd.read_csv("/content/sales_data_validated.csv").set_index("Row ID")

with open("/content/splits.json") as f:
    splits = json.load(f)

def create_models(seed):
    return {
        "LinearRegression": LinearRegression(),
        "RandomForest": RandomForestRegressor(
            n_estimators=100, random_state=seed, n_jobs=-1
        ),
        "XGBoost": XGBRegressor(
            n_estimators=100, max_depth=6, learning_rate=0.1,
            random_state=seed, n_jobs=-1,
            objective="reg:squarederror"
        )
    }

all_results = []

for seed in [42, 43, 44, 45, 46]:
    print(f"Running seed {seed}...")

    split = splits[f"order_grouped_seed{seed}"]

    train_df = df.loc[split["train_row_ids"]].sort_index().copy()
    test_df = df.loc[split["test_row_ids"]].sort_index().copy()

    X_train = train_df.drop(columns=[TARGET])
    y_train = train_df[TARGET]
    X_test = test_df.drop(columns=[TARGET])
    y_test = test_df[TARGET]
    groups = train_df["Order ID"].to_numpy()

    names = list(create_models(seed).keys())
    oof = np.zeros((len(X_train), len(names)))
    cv = GroupKFold(n_splits=5)

    for fit_idx, val_idx in cv.split(X_train, y_train, groups):
        for j, (name, model) in enumerate(create_models(seed).items()):
            pipe = make_pipeline(model, use_shipping_cost=False)
            pipe.fit(X_train.iloc[fit_idx], y_train.iloc[fit_idx])
            oof[val_idx, j] = pipe.predict(X_train.iloc[val_idx])

    meta = Ridge(alpha=1.0)
    meta.fit(oof, y_train)

    test_preds = np.zeros((len(X_test), len(names)))

    for j, (name, model) in enumerate(create_models(seed).items()):
        pipe = make_pipeline(model, use_shipping_cost=False)
        pipe.fit(X_train, y_train)
        test_preds[:, j] = pipe.predict(X_test)

    predictions = {
        "RandomForest": test_preds[:, names.index("RandomForest")],
        "StackingRegressor": meta.predict(test_preds)
    }

    for name, pred in predictions.items():
        all_results.append({
            "seed": seed,
            "model": name,
            "MAE": mean_absolute_error(y_test, pred),
            "RMSE": np.sqrt(mean_squared_error(y_test, pred)),
            "R2": r2_score(y_test, pred)
        })

results_df = pd.DataFrame(all_results).round(4)
display(results_df)

summary = results_df.groupby("model")[["MAE", "RMSE", "R2"]].agg(
    ["mean", "std"]
).round(4)

print("Five-seed summary:")
display(summary)

results_df.to_csv(
    "/content/reports/stacking_five_seeds_no_shipping_sorted.csv",
    index=False
)

print("Saved corrected five-seed results.")
```

```python

import json
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load data and the saved split
df = pd.read_csv("/content/sales_data_validated.csv")

with open("/content/splits.json") as f:
    saved_splits = json.load(f)

split = saved_splits["order_grouped_seed42"]

# Match saved Row ID values to the actual dataset rows
df = df.set_index("Row ID", drop=False)

train_df = df.loc[split["train_row_ids"]].copy()
test_df = df.loc[split["test_row_ids"]].copy()

# Define the features used in the experiment
categorical = [
    "Category", "Sub-Category", "Segment", "Market",
    "Region", "Ship Mode", "Order Priority"
]
numeric = ["Sales", "Quantity", "Discount"]

def evaluate_features(numeric_columns, label):
    features = numeric_columns + categorical

    preprocessing = ColumnTransformer([
        ("num", "passthrough", numeric_columns),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical)
    ])

    model = Pipeline([
        ("preprocessing", preprocessing),
        ("model", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ])

    model.fit(train_df[features], train_df["Profit"])
    predictions = model.predict(test_df[features])

    return {
        "Scenario": label,
        "Features used": len(features),
        "MAE": mean_absolute_error(test_df["Profit"], predictions),
        "RMSE": np.sqrt(mean_squared_error(test_df["Profit"], predictions)),
        "R2": r2_score(test_df["Profit"], predictions)
    }

results = pd.DataFrame([
    evaluate_features(numeric, "All standard features"),
    evaluate_features(
        ["Sales", "Quantity"],
        "Without Discount"
    )
]).round(4)

display(results)

results.to_csv(
    "/content/reports/missing_feature_robustness.csv",
    index=False
)

print("Saved missing_feature_robustness.csv")
print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))
print("Row ID overlap:", len(set(train_df.index) & set(test_df.index)))
```

```python

import json
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = pd.read_csv("/content/sales_data_validated.csv")

with open("/content/splits.json") as f:
    splits = json.load(f)

categorical = [
    "Category", "Sub-Category", "Segment", "Market",
    "Region", "Ship Mode", "Order Priority"
]
numeric = ["Sales", "Quantity", "Discount"]
features = numeric + categorical

rows = []

for seed in [42, 43, 44, 45, 46]:
    split = splits[f"order_grouped_seed{seed}"]

    indexed = df.set_index("Row ID", drop=False)
    train = indexed.loc[split["train_row_ids"]].copy()
    test = indexed.loc[split["test_row_ids"]].copy()

    for fraction in [0.50, 0.75, 1.00]:
        sample = train.sample(
            frac=fraction, random_state=seed
        )

        preprocessing = ColumnTransformer([
            ("num", "passthrough", numeric),
            ("cat", OneHotEncoder(handle_unknown="ignore"), categorical)
        ])

        model = Pipeline([
            ("preprocessing", preprocessing),
            ("model", RandomForestRegressor(
                n_estimators=100,
                random_state=seed,
                n_jobs=-1
            ))
        ])

        model.fit(sample[features], sample["Profit"])
        pred = model.predict(test[features])

        rows.append({
            "seed": seed,
            "training_fraction": fraction,
            "training_rows": len(sample),
            "MAE": mean_absolute_error(test["Profit"], pred),
            "RMSE": np.sqrt(mean_squared_error(test["Profit"], pred)),
            "R2": r2_score(test["Profit"], pred)
        })

results = pd.DataFrame(rows)

summary = results.groupby("training_fraction").agg(
    training_rows_mean=("training_rows", "mean"),
    MAE_mean=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    RMSE_mean=("RMSE", "mean"),
    RMSE_std=("RMSE", "std"),
    R2_mean=("R2", "mean"),
    R2_std=("R2", "std")
).round(4)

display(summary)

results.to_csv(
    "/content/reports/training_size_robustness_five_seeds.csv",
    index=False
)
summary.to_csv(
    "/content/reports/training_size_robustness_summary.csv"
)

print("Saved both five-seed robustness reports.")
```

```python

import pandas as pd
import numpy as np
from scipy.stats import ttest_rel, t

path = "/content/reports/stacking_five_seeds_no_shipping_sorted.csv"
results = pd.read_csv(path)

print("Columns:", results.columns.tolist())

required = {"seed", "model", "MAE", "RMSE", "R2"}
if not required.issubset(results.columns):
    raise ValueError(
        f"Expected columns {required}; found {set(results.columns)}"
    )

wide = results.pivot(index="seed", columns="model",
                     values=["MAE", "RMSE", "R2"])

required_models = {"RandomForest", "StackingRegressor"}
if not required_models.issubset(set(results["model"])):
    raise ValueError("RandomForest or StackingRegressor is missing.")

rows = []

for metric in ["MAE", "RMSE", "R2"]:
    rf = wide[metric]["RandomForest"]
    stack = wide[metric]["StackingRegressor"]

    # Positive improvement means stacking is better.
    improvement = rf - stack if metric in ["MAE", "RMSE"] else stack - rf

    n = improvement.notna().sum()
    mean_diff = improvement.mean()
    std_diff = improvement.std(ddof=1)

    if n > 1:
        margin = t.ppf(0.975, n - 1) * std_diff / np.sqrt(n)
        p_value = ttest_rel(rf, stack).pvalue
    else:
        margin = np.nan
        p_value = np.nan

    rows.append({
        "Metric": metric,
        "Mean improvement": mean_diff,
        "95% CI lower": mean_diff - margin,
        "95% CI upper": mean_diff + margin,
        "Paired t-test p-value": p_value,
        "Seeds": n
    })

summary = pd.DataFrame(rows).round(4)
display(summary)

summary.to_csv(
    "/content/reports/stacking_statistical_tests.csv",
    index=False
)
print("Saved stacking_statistical_tests.csv")
```

```python

import json
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

report_dir = Path("/content/reports")

df = pd.read_csv("/content/sales_data_validated.csv")
pred = pd.read_csv(report_dir / "final_predictions.csv")

with open("/content/splits.json", "r") as f:
    saved_splits = json.load(f)

split = saved_splits["order_grouped_seed42"]

# Match saved Row IDs to their positions in the validated dataset
row_positions = pd.Series(
    np.arange(len(df)),
    index=df["Row ID"]
)

test_positions = row_positions.loc[
    split["test_row_ids"]
].to_numpy()

test_df = df.iloc[test_positions].reset_index(drop=True)

# Verify prediction rows match the intended test rows
actual_match = np.allclose(
    test_df["Profit"].to_numpy(),
    pred["Actual_Profit"].to_numpy(),
    equal_nan=True
)

print("Dataset rows:", len(df))
print("Test rows:", len(test_df))
print("Prediction rows:", len(pred))
print("Actual-profit alignment verified:", actual_match)

if actual_match and len(test_df) == len(pred):
    print("\nAlignment verified. Ready for segment-wise analysis.")
else:
    print("\nAlignment mismatch detected. Do not calculate segment metrics yet.")
```

```python

import pandas as pd
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Attach verified predictions to the matching test rows
analysis_df = test_df.copy()
analysis_df["Actual_Profit"] = pred["Actual_Profit"].to_numpy()
analysis_df["Predicted_Profit"] = pred["Predicted_Profit"].to_numpy()
analysis_df["Absolute_Error"] = (
    analysis_df["Actual_Profit"] - analysis_df["Predicted_Profit"]
).abs()

group_columns = ["Segment", "Category", "Region"]
results = []

for column in group_columns:
    for group_name, group in analysis_df.groupby(column, dropna=False):
        actual = group["Actual_Profit"]
        predicted = group["Predicted_Profit"]

        results.append({
            "Grouping": column,
            "Group": str(group_name),
            "Rows": len(group),
            "MAE": mean_absolute_error(actual, predicted),
            "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
            "R2": r2_score(actual, predicted) if len(group) > 1 else np.nan
        })

segment_results = pd.DataFrame(results).round(4)

print("SEGMENT-WISE MODEL PERFORMANCE")
display(segment_results)

output_path = "/content/reports/segment_wise_errors.csv"
segment_results.to_csv(output_path, index=False)

print("Saved:", output_path)
print("\nLowest MAE groups within each grouping:")
for column in group_columns:
    subset = segment_results[segment_results["Grouping"] == column]
    best = subset.loc[subset["MAE"].idxmin()]
    print(f"{column}: {best['Group']} | MAE = {best['MAE']}")
```

```python

import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

df = pd.read_csv("/content/sales_data_validated.csv")

category_filter = widgets.Dropdown(
    options=["All"] + sorted(df["Category"].dropna().unique().tolist()),
    value="All",
    description="Category:"
)

region_filter = widgets.Dropdown(
    options=["All"] + sorted(df["Region"].dropna().unique().tolist()),
    value="All",
    description="Region:"
)

output = widgets.Output()

def update_dashboard(change=None):
    filtered = df.copy()

    if category_filter.value != "All":
        filtered = filtered[
            filtered["Category"] == category_filter.value
        ]

    if region_filter.value != "All":
        filtered = filtered[
            filtered["Region"] == region_filter.value
        ]

    with output:
        clear_output(wait=True)

        sales = filtered["Sales"].sum()
        profit = filtered["Profit"].sum()
        orders = filtered["Order ID"].nunique()
        profit_per_order = profit / orders if orders else 0

        display(HTML(f"""
        <h3>Filtered Business Summary</h3>
        <p><b>Total Sales:</b> ${sales:,.2f}</p>
        <p><b>Total Profit:</b> ${profit:,.2f}</p>
        <p><b>Unique Orders:</b> {orders:,}</p>
        <p><b>Profit per Order:</b> ${profit_per_order:,.2f}</p>
        <p><b>Records:</b> {len(filtered):,}</p>
        """))

        chart_data = (
            filtered.groupby("Category", as_index=False)
            .agg(Profit=("Profit", "sum"))
            .sort_values("Profit", ascending=False)
        )

        fig = px.bar(
            chart_data,
            x="Category",
            y="Profit",
            color="Category",
            title="Profit by Category — Filtered Results"
        )
        fig.show()

category_filter.observe(update_dashboard, names="value")
region_filter.observe(update_dashboard, names="value")

display(widgets.HBox([category_filter, region_filter]))
display(output)

update_dashboard()
```

```python

import pandas as pd
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output
from sklearn.ensemble import RandomForestRegressor
from common import grouped_split, make_pipeline, TARGET, NUMERIC, CATEGORICAL

# Load the validated dataset
df = pd.read_csv("/content/sales_data_validated.csv")

# Use the same order-grouped split approach as the experiments
train_idx, test_idx = grouped_split(df, 42)
train_df = df.iloc[train_idx].copy()

# Exclude Shipping Cost to match the selected experiment
features = NUMERIC + CATEGORICAL

model = make_pipeline(
    RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),
    use_shipping_cost=False
)

model.fit(train_df[features], train_df[TARGET])

print("Model trained successfully.")
print("Training rows:", len(train_df))
print("Features used:", features)

# Create input fields
sales_input = widgets.FloatText(
    value=200.0, description="Sales:", style={"description_width": "initial"}
)
quantity_input = widgets.IntText(
    value=2, description="Quantity:", style={"description_width": "initial"}
)
discount_input = widgets.FloatSlider(
    value=0.1, min=0, max=0.8, step=0.05,
    description="Discount:", style={"description_width": "initial"}
)

def make_dropdown(column):
    values = sorted(df[column].dropna().astype(str).unique().tolist())
    return widgets.Dropdown(
        options=values,
        value=values[0],
        description=f"{column}:",
        style={"description_width": "initial"},
        layout=widgets.Layout(width="350px")
    )

segment_input = make_dropdown("Segment")
category_input = make_dropdown("Category")
subcategory_input = make_dropdown("Sub-Category")
market_input = make_dropdown("Market")
region_input = make_dropdown("Region")
shipmode_input = make_dropdown("Ship Mode")
priority_input = make_dropdown("Order Priority")

predict_button = widgets.Button(
    description="Predict Profit",
    button_style="success",
    icon="check"
)
result_output = widgets.Output()

def predict_profit(button):
    with result_output:
        clear_output(wait=True)

        row = {
            "Sales": sales_input.value,
            "Quantity": quantity_input.value,
            "Discount": discount_input.value,
            "Segment": segment_input.value,
            "Category": category_input.value,
            "Sub-Category": subcategory_input.value,
            "Market": market_input.value,
            "Region": region_input.value,
            "Ship Mode": shipmode_input.value,
            "Order Priority": priority_input.value
        }

        if sales_input.value < 0 or quantity_input.value < 1:
            print("Please enter non-negative sales and a quantity of at least 1.")
            return

        input_df = pd.DataFrame([row])[features]
        prediction = model.predict(input_df)[0]

        print(f"Predicted profit: {prediction:,.2f}")
        print("This is a model estimate, not a guaranteed actual profit.")

predict_button.on_click(predict_profit)

display(widgets.HTML("<h2>Retail Profit Prediction</h2>"))
display(widgets.HBox([sales_input, quantity_input]))
display(discount_input)
display(widgets.HBox([segment_input, category_input]))
display(widgets.HBox([subcategory_input, market_input]))
display(widgets.HBox([region_input, shipmode_input]))
display(priority_input)
display(predict_button, result_output)
```
